```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'..\dataset\CleanedData\titanic.csv')
```

```python
d = {True:'Survive',
    False:'Not Survive'}
df['Survived'] =df['Survived'].map(d)
```

```python
x = df.groupby(['Survived', 'Gender'])['PassengerId'].count()
```

```python
x.index
```

#### Piechart

```python
plt.pie(x, 
        labels= ('dead female','dead male','female','male'),
        autopct='%1.1f%%',
        explode=[0,0,0,0.3],
        shadow=True,
        startangle= 90,
        counterclock=False,
        radius=1.5,
        colors=['pink', 'purple', 'red', 'green'],
        labeldistance=1.05)
```
