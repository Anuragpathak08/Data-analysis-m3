#### Subplot

```python
plt.subplot(2 , 2, 1)
plt.plot([1,2,3], [3,4,5])

plt.subplot(2 , 2, 2)
plt.plot([1,2,3], [3,4,5])

plt.subplot(2 , 2, 3)
plt.plot([1,2,3], [3,4,5])

plt.subplot(2 , 2, 4)
plt.plot([1,2,3], [3,4,5])
plt.yticks([0,2,4,6], ['abcd', 'efgh', 'Yash', 'Raju Ji'])
plt.tight_layout()
```
