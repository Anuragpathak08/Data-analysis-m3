#### Renaming the column

```python
rename = {'listed_in(type)': 'Type',
          'listed_in(city)':'city',
          'approx_cost(for two people)': 'cost_of_2'}

zomato = zomato.rename(columns=rename)
```

#### Remove Duplicated

```python
zomato.duplicated().sum()
```

#### Handeling the nan value

```python
zomato.isna().sum()
```

```python
zomato = zomato.dropna(subset=['location'])
zomato['location'].isnull().sum()
```

Before filling the na in rating with median we need to clean the rating and convert the dtype to int

```python
(zomato['rate'] == 'NEW').sum()
```

```python
zomato['rate'] = zomato['rate'].str.replace('/5', '').replace('NEW', '0').replace('-', '0').astype('f4')
```

##### fill the nan of rate

```python
zomato['rate'] = zomato['rate'].fillna(0)
```

**removed the resturant type null value**

```python
zomato = zomato.dropna(subset=['rest_type'])
```

**cost_of_2 is a str we need to clean and change the dtype**

```python
zomato['cost_of_2'] = zomato['cost_of_2'].str.replace(',', '').astype('f4')
```

**cost_of_2 nan value will be filled with mean**

```python
co2 = zomato['cost_of_2'].mean()
zomato['cost_of_2'] = zomato['cost_of_2'].fillna(co2)
```

```python
zomato['cost_of_2'] = zomato['cost_of_2'].astype('int16')
```

#### changing the data in  a column

```python
zomato['phone'] = zomato['phone'].notna()
```

```python
zomato['cuisines'] = zomato['cuisines'].ffill()
```

```python
zomato.isna().sum()
```

#### Removing the unwanted column

```python
zomato = zomato.drop(columns=['url', 'address', 'dish_liked', 'menu_item', 'reviews_list'])
```
