### Functions of NumPy

NumPy provides many built-in functions for creating arrays, generating values, and performing mathematical operations.

These functions are usually called using the `np` alias, for example `np.arange()`, `np.zeros()`, and `np.random.rand()`.

#### NumPy Function vs Array Method

A **NumPy function** is called from the NumPy module:

```python
np.arange(10)
np.sum(arr)
```

An **array method** is called directly on an array object:

```python
arr.sum()
arr.reshape(2, 5)
```

Short difference:

- `np.function()` belongs to the NumPy module.
- `arr.method()` belongs to a specific array object.
- Some operations can be done both ways, for example `np.sum(arr)` and `arr.sum()`.

````python
import numpy as np
````

### NumPy Functions

#### arange

`np.arange()` returns evenly spaced values within a given interval.

Basic syntax:

```python
np.arange(start, stop, step)
```

- `start` is included
- `stop` is excluded
- `step` is the difference between consecutive values

````python
np.arange(10)
````

When only one argument is given, NumPy generates values from `0` up to one less than that value.

````python
np.arange(10, -10, -1)
````

We can also use a negative step value to generate values in reverse order.

````python
np.arange(1, 101, 2).reshape(2, 5, 5)
````

Here, `np.arange(1, 101, 2)` generates 50 odd numbers from 1 to 99.

Then `.reshape(2, 5, 5)` reshapes those 50 values into a 3D array with shape `(2, 5, 5)`.

````python
np.arange(0, 20, 1.25)
````

`np.arange()` also supports float step values.

However, because of floating-point representation, values may sometimes have small precision issues.

#### zeros

`np.zeros()` creates an array filled with zeros.

By default, the dtype is `float64`.

To define a multidimensional shape, we can use a list or tuple.

````python
np.zeros(3) # array([0., 0., 0.])
np.zeros([5, 3]) # float
np.zeros((5, 3)).astype('int8') # int

# to define a shape we can only use list or tuple
````

In the cell above, three expressions are written, but Jupyter displays only the last output.

- `np.zeros(3)` creates a 1D array of three float zeros
- `np.zeros([5, 3])` creates a 2D array of float zeros
- `.astype('int8')` converts the zero array to integer dtype

#### ones

`np.ones()` creates an array filled with ones.

By default, the dtype is `float64`.

````python
np.ones(5)
np.ones((2, 9))
np.ones((2, 9)).astype('int8')
````

- `np.ones(5)` creates a 1D array of five float ones
- `np.ones((2, 9))` creates a 2D array of float ones
- `.astype('int8')` converts the ones to integer dtype

#### linspace

`np.linspace()` returns evenly spaced numbers over a specified interval.

Basic syntax:

```python
np.linspace(start, stop, num)
```

- `start` is included
- `stop` is included by default
- `num` is the number of values to generate

````python
np.linspace(1, 50)
np.linspace(1, 26, 100)
````

- `np.linspace(1, 50)` generates 50 values between 1 and 50 by default
- `np.linspace(1, 26, 100)` generates 100 values between 1 and 26

#### identity

`np.identity()` creates a square identity matrix.

An identity matrix has ones on the main diagonal and zeros everywhere else.

````python
np.identity(3, dtype='int')
````

#### random.random

`np.random.random()` generates random floating-point values between 0 and 1.

- 0 is included
- 1 is excluded

````python
np.random.random() 
np.random.random(3) 
np.random.random(3) * 100
````

Here, three random values between 0 and 1 are generated and then multiplied by 100.

Random outputs may be different every time the cell is executed.

#### random.randint

`np.random.randint()` generates random integers.

Basic syntax:

```python
np.random.randint(low, high, size)
```

- `low` is included
- `high` is excluded
- `size` defines how many values to generate

````python
np.random.randint(10)
np.random.randint(10, 20)
np.random.randint(10, 2000, 90)
````

- `np.random.randint(10)` gives one random integer from 0 to 9
- `np.random.randint(10, 20)` gives one random integer from 10 to 19
- `np.random.randint(10, 2000, 90)` gives 90 random integers from 10 to 1999

#### random.rand

`np.random.rand()` generates random floating-point values between 0 and 1 with a given shape.

Unlike `np.random.random()`, the dimensions are passed as separate arguments.

````python
np.random.rand(2, 3)
````

- `np.arange()` creates evenly spaced values using a step
- `np.linspace()` creates evenly spaced values using a count
- `np.zeros()` creates arrays filled with zeros
- `np.ones()` creates arrays filled with ones
- `np.identity()` creates an identity matrix
- `np.random.random()` generates random floats between 0 and 1
- `np.random.randint()` generates random integers
- `np.random.rand()` generates random floats with shape passed as separate arguments
- NumPy functions are called using `np.function()`
- Array methods are called using `array.method()`
