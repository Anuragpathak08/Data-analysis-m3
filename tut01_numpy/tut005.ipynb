#### arange

```python
np.arange(10,-10, -1)
```

#### zeros

```python
np.zeros(3) # array([0., 0., 0.])
np.zeros([5,3]) # float
np.zeros((5,3)).astype('int8') # int

# to define a shape we can only use list or tuple
```

#### ones

```python
np.ones(5)
np.ones((2, 9))
np.ones((2, 9)).astype('int8')
```

```python
np.linspace(1, 50)
np.linspace(1, 26,100)
```

```python
np.identity(3, dtype= 'int')
```
