### Creating NumPy Arrays

NumPy arrays are created using `np.array()`. This function accepts an input object and tries to convert it into a NumPy `ndarray`.

The important concept is how NumPy interprets the input:

- If the input is an ordered sequence-like object, such as a list, tuple, or an existing NumPy array, NumPy usually treats each item inside it as a separate array element.
- If the input is not treated as a sequence-like collection, such as a set, dictionary, single number, boolean, complex number, or string, NumPy often treats the whole object as one element and creates a 0-dimensional array.

In data analysis, lists and NumPy arrays are the most commonly supported and reliable inputs for array creation.

### Syntax

```python
np.array(object)
```

### How NumPy decides what becomes an array element

NumPy works best with ordered collections where each item can be accessed in a fixed sequence.

Examples of inputs that are expanded into elements:

- Python list: `[1, 2, 3]`
- Python tuple: `(1, 2, 3)`
- Existing NumPy array: `np.array([1, 2, 3])`

Examples of inputs that are often treated as one single element:

- Set: `{1, 2, 3}`
- Dictionary: `{'a': 1}`
- Single number: `10`
- Boolean: `True`
- String: `'hello world'`

This happens because NumPy expects a predictable sequence of elements. Objects like sets and dictionaries do not behave like regular ordered numeric sequences for array creation.

#### Array using list

A Python list is one of the most common ways to create a NumPy array. Each value inside the list becomes an element of the array.

````python
arr1 = np.array([1233, 132, 3, 243, 43, 5])
arr1
````

Here, the list `[1233, 132, 3, 243, 43, 5]` is converted into a 1-dimensional NumPy array.

Since all values are integers, NumPy infers an integer dtype automatically.

#### Array from an existing NumPy array

If we pass an existing NumPy array into `np.array()`, NumPy creates a new array object from it.

````python
arr1_from_array = np.array(arr1)
arr1_from_array
````

This is useful when we want to create a separate array object from an existing array.

#### Array using tuple

A tuple is also an ordered sequence, so NumPy can convert its elements into an array.

````python
arr2 = np.array((12, 87, 3, 76, 43))
arr2
````

Even though lists are more commonly used, tuples also work because they provide a fixed ordered collection of values.

#### Array using set

A Python set is not treated like a normal ordered sequence by NumPy. Therefore, instead of converting each set item into an array element, NumPy stores the whole set as one object.

````python
arr3 = np.array({13, 2, 4})
arr3
````

````python
arr3.shape # ()
arr3.ndim  # 0
````

The output shows `dtype=object`, which means NumPy stored the Python set itself as a single object.

The shape is `()` and the dimension is `0`, so this is a 0-dimensional array.

If we want each element of the set to become an array element, we should first convert the set into a list:

```python
np.array(list({13, 2, 4}))
```

However, sets are unordered, so the order may change.

#### Array using string

When a plain string is passed to `np.array()`, NumPy does not split it into characters. It treats the full string as one value.

````python
arr4 = np.array('hello world')
arr4
````

Here, `'hello world'` becomes a single array element with Unicode dtype `<U11` because the string length is 11.

If we want each character as a separate array element, we can convert the string into a list first:

```python
np.array(list('hello world'))
```

#### Array using single values

If we pass a single Python value, NumPy creates a 0-dimensional array containing that value.

````python
np.array(10)
np.array(20.3)
np.array(20 + 4j)
np.array(True)
````

In this cell, only the last expression is displayed, which is `array(True)`.

Each single value becomes a scalar array *(array with one value)*:

- `np.array(10)` creates an integer scalar array
- `np.array(20.3)` creates a float scalar array
- `np.array(20 + 4j)` creates a complex scalar array
- `np.array(True)` creates a boolean scalar array

#### Boolean arrays and mixed data types

When a list contains mixed data types, NumPy tries to convert all values into a common dtype.

````python
arr = np.array([True, False, True, True])
arr
````

If all values are booleans, NumPy creates a boolean array.

````python
arr = np.array([True, False, 10, False])
arr
````

When booleans are mixed with integers, NumPy converts booleans into integers:

- `True` becomes `1`
- `False` becomes `0`

So the final array becomes an integer array.

````python
arr = np.array([True, 10, 230, 3, 10.234])
arr
````

When integers and floats are mixed, NumPy promotes the array to float dtype because float can safely represent integer values.

````python
arr = np.array([True, 10, 21.3, 224 + 2j])
arr
````

When complex numbers are present, NumPy promotes the array to complex dtype because complex is the broadest numeric type in this example.

Numpy array follows a prority to coverts all the values into a common type 
```plaintext
      bool       (Lowest priority)
        │
        ▼
       int
        │
        ▼
      float
        │
        ▼
     complex
        │
        ▼
      string  
        │
        ▼
      object     (Highest priority)
```
