### Indexing, Slicing, Fancy Indexing and Boolean Indexing

NumPy arrays support powerful ways to access and select data:

- **Indexing**: selecting one element
- **Slicing**: selecting a continuous range of elements
- **Fancy indexing**: selecting arbitrary elements using arrays of indices
- **Boolean indexing**: selecting elements based on a condition

For multidimensional arrays, each dimension is called an **axis**.

- 1D array has axis 0 only
- 2D array has axis 0 as rows and axis 1 as columns
- 3D array has axis 0 as blocks, axis 1 as rows, and axis 2 as columns

````python
import numpy as np
````

````python
arr1 = np.random.randint(10, 30, 15)
arr2 = np.random.randint(10, 40, 30).reshape(5, 6)
arr3 = np.random.randint(10, 50, 40).reshape(5, 2, 4)
````

````python
arr1
````

````python
arr2
````

````python
arr3
````

### Indexing

Indexing is used to select a single element.

Important points:

- Indexing starts from 0
- Negative indexing counts from the end
- For multidimensional arrays, we can use either chained indexing or comma indexing
- Comma indexing is usually preferred in NumPy

````python
arr1[2]
arr1[-1]
arr1[-15]
arr1[14]
````

For a 1D array:

- `arr1[2]` gives the element at index 2
- `arr1[-1]` gives the last element
- `arr1[-15]` gives the first element because the array has 15 elements

````python
print(arr2[0][0])
print(arr2[0][-1])
print(arr2[0, 3])
````

For a 2D array:

- `arr2[0][0]` selects row 0, then column 0
- `arr2[0][-1]` selects row 0, then the last column
- `arr2[0, 3]` directly selects row 0 and column 3

The comma style `arr2[row, column]` is the recommended NumPy style.

````python
print(arr3[0][0][0])
print(arr3[0, 0, 0])
````

For a 3D array:

- `arr3[0][0][0]` selects block 0, row 0, column 0
- `arr3[0, 0, 0]` does the same thing using comma indexing

### Slicing

Slicing selects a range of elements using the syntax:

```python
start:stop:step
```

- `start` is included
- `stop` is excluded
- `step` defines the gap between elements
- If `start` is omitted, it defaults to 0
- If `stop` is omitted, it defaults to the end
- If `step` is omitted, it defaults to 1

#### 1D slicing

````python
arr1[2:7]
````

`arr1[2:7]` selects indices 2, 3, 4, 5, and 6.

Index 7 is excluded.

````python
arr1[:5]
````

````python
arr1[::2]
````

`arr1[::2]` selects every second element from the whole array.

````python
arr1[::-1]
````

`arr1[::-1]` reverses the array.

````python
arr1[-5:]
````

`arr1[-5:]` selects the last five elements.

#### 2D slicing and axes

For a 2D array:

```python
arr2[row_slice, column_slice]
```

- The part before the comma is axis 0, which means rows
- The part after the comma is axis 1, which means columns
- `:` means select everything along that axis

````python
arr2[0:3, :]
````

`arr2[0:3, :]` slices axis 0.

It selects rows 0, 1, and 2, and all columns.

````python
arr2[:, 0:3]
````

`arr2[:, 0:3]` slices axis 1.

It selects all rows, and columns 0, 1, and 2.

````python
arr2[0:3, 0:3]
````

`arr2[0:3, 0:3]` slices both axes.

It selects rows 0, 1, and 2, and columns 0, 1, and 2.

````python
arr2[::2, :]
````

`arr2[::2, :]` uses a step value on axis 0.

It selects every second row.

````python
arr2[:, ::2]
````

`arr2[:, ::2]` uses a step value on axis 1.

It selects every second column.

#### 3D slicing and axes

For a 3D array:

```python
arr3[block_slice, row_slice, column_slice]
```

- axis 0: blocks
- axis 1: rows
- axis 2: columns

````python
arr3[0, :, :]
````

`arr3[0, :, :]` selects block 0, all rows, and all columns.

````python
arr3[:, 0, :]
````

`arr3[:, 0, :]` selects all blocks, row 0, and all columns.

````python
arr3[:, :, 0]
````

`arr3[:, :, 0]` selects all blocks, all rows, and column 0.

````python
arr3[..., 0]
````

`arr3[..., 0]` is a shortcut using `...`.

Here, `...` means all previous axes, so this is equivalent to `arr3[:, :, 0]`.

````python
arr3[1:4, :, 1:3]
````

`arr3[1:4, :, 1:3]` slices all three axes:

- blocks 1, 2, and 3
- all rows
- columns 1 and 2

### Fancy Indexing

Fancy indexing is used when we want to select arbitrary elements, not necessarily a continuous slice.

Instead of using `start:stop:step`, we pass a list or array of indices.

````python
arr1[[0, 5, 1, 3, 5, 5, 5, 5]]
````

Here, NumPy selects the elements at indices 0, 5, 1, 3, 5, 5, 5, and 5.

The same index can also be repeated.

````python
arr2[[0, 3, 1], [1]]
````

In this example:

- row indices are `[0, 3, 1]`
- column indices are `[1]`

NumPy broadcasts `[1]` to match the row index list, so it selects:

- `arr2[0, 1]`
- `arr2[3, 1]`
- `arr2[1, 1]`

````python
arr3[[0, 4, 3], [0]]
````

In this 3D example:

- block indices are `[0, 4, 3]`
- row indices are `[0]`
- the column axis is not specified, so all columns are selected

So NumPy selects:

- `arr3[0, 0, :]`
- `arr3[4, 0, :]`
- `arr3[3, 0, :]`

### Boolean Indexing

Boolean indexing selects elements based on a condition.

It is very useful for filtering data.

````python
arr1[arr1 > 20]
````

`arr1[arr1 > 20]` returns only the values from `arr1` that are greater than 20.

````python
arr2[arr2 > 30]
````

`arr2[arr2 > 30]` returns all elements of `arr2` that are greater than 30.

The result is flattened into a 1D array.

````python
arr2[arr2[:, 0] > 15]
````

Here:

- `arr2[:, 0]` selects the first column
- `arr2[:, 0] > 15` creates a boolean mask for rows
- `arr2[arr2[:, 0] > 15]` selects only the rows where the first column is greater than 15

### View vs Copy

- Basic slicing usually returns a **view** of the original array.
- Fancy indexing and boolean indexing usually return a **copy**.

If you want to make sure that you have an independent copy, use `.copy()`.

### Key Takeaways

- Indexing selects single elements
- Slicing selects ranges using `start:stop:step`
- In multidimensional arrays, each axis can be sliced separately
- `:` means select everything along that axis
- `...` can be used to represent all previous axes
- Fancy indexing selects arbitrary elements using index arrays
- Boolean indexing filters elements using conditions
- Basic slicing often returns a view, while fancy/boolean indexing usually returns a copy
