### Methods of numpy array

NumPy arrays (`ndarray`) come with a rich set of built-in methods for performing mathematical, statistical, and structural operations. Using these methods directly on the array object is highly optimized and often preferred over using standalone NumPy functions.

#### sum

The `.sum()` method calculates the sum of all elements in the array. 
- For 1D arrays, it returns the total sum.
- For multi-dimensional arrays, you can specify the `axis` parameter to sum along rows (`axis=0`) or columns (`axis=1`).

```python
arr1 = np.array([12,4,3,5,6,4])
arr1.sum()
```

```python
arr2 = np.array(['hello!!', 'hi', 'how', 'are', 'you'])
# arr2.sum() #error

arr3 = np.array([10+2j, 4j, 45 + 1j])
arr3.sum()

arr4 = np.array([13.3, 41.5, 6.5, 1.7, 8.9])
arr4.sum()

arr5 = np.array([True , False , True , True , False])
arr5.sum()
```

#### prod

The `.prod()` method computes the product of all array elements. Like `sum`, it can also accept an `axis` argument to calculate the product along a specific dimension.

```python
arr1.prod()
```

#### mean

The `.mean()` method calculates the arithmetic average of the array elements. It sums all the elements and divides by the total number of elements. This method is only applicable to numeric data types (integers, floats, complex).

```python
arr1.mean()
# arr2.mean() #string doesn't have mean
arr3.mean()
arr4.mean()
arr5.mean()
```

#### median

The median is the middle value of a sorted dataset. Unlike `sum` or `mean`, `median` is typically called as a NumPy function (`np.median()`) rather than an array method. It effectively finds the central tendency of the data, ignoring extreme outliers.

```python
arr = np.array([9,2,1,6,4,7,11,8,14])
np.median(arr)
```

#### std

The `.std()` method calculates the **standard deviation** of the array elements. It measures how spread out the numbers are from the mean. By default, it calculates the population standard deviation, but you can use the `ddof` (Delta Degrees of Freedom) parameter to calculate the sample standard deviation.

```python
arr6 = np.array([0,5,20, 15, 3, 7, 9])
arr6.std()
```

#### cumsum

The `.cumsum()` method returns the **cumulative sum** of the elements. Each element in the resulting array is the sum of all previous elements plus the current element. This is useful for calculating running totals.

```python
arr1
```

```python
arr1.cumsum()
```

#### cumprod

The `.cumprod()` method returns the **cumulative product** of the elements. Similar to `cumsum`, each element in the output array is the product of all preceding elements and the current element.

```python
arr1.cumprod()
```

#### reshape

The `.reshape()` method changes the shape (dimensions) of an array without altering its underlying data. 
- The total number of elements in the new shape must exactly match the original array's size.
- You can pass `-1` for one of the dimensions, and NumPy will automatically infer the correct size based on the remaining dimensions and the total number of elements.

```python
arr = np.array([1,4,3,5,6,5,3,4,34,54,54,6,18,56,5,7,67,6])
arr.size #18
arr.reshape(6,3) # product of 6*3 =18 hence we can reshape
# arr.reshape(6,2) # product of 6*2 !=18 hence we cannot reshape
```

NumPy supports a comprehensive set of data types to handle various kinds of data efficiently. Here is a breakdown of the primary data type categories:

1. **Integer Types**
    - Signed integers: `int8`, `int16`, `int32`, `int64`
    - Unsigned integers: `uint8`, `uint16`, `uint32`, `uint64`

2. **Floating-Point & Complex Types**
    - Floats: `float16`, `float32`, `float64`
    - Complex numbers: `complex64`, `complex128`

3. **Boolean Type**
    - `bool` (converts zero/empty to `False` and non-zero to `True`)

4. **String & Text Types**
    - Fixed-width ASCII/Bytes: `string_` (e.g., `S10`)
    - Unicode: `unicode_` (e.g., `U10`)

5. **Advanced & Flexible Types**
    - Python Objects: `object` (allows the array to store arbitrary Python objects)
    - Datetime types: `datetime64` and `timedelta64`
