### Types of NumPy Arrays

NumPy arrays can have different numbers of dimensions. The number of dimensions is called the **ndim** of the array.

- **1D array**: one axis, like a simple list of values
- **2D array**: two axes, like a table with rows and columns
- **3D array**: three axes, like a collection of 2D tables


````python
import numpy as np
````

````python
np.array([1, 3, 34, 5, 23, 24])
````

#### Types of array

NumPy arrays are often classified by their dimension:

| Type | Meaning | Example shape |
|---|---|---|
| 1D array | single sequence | `(4,)` |
| 2D array | rows and columns | `(2, 3)` |
| 3D array | stack of 2D arrays | `(2, 2, 3)` |

#### 1D array

A 1D array has only one axis. It is similar to a normal Python list, but it is stored as a NumPy array.

````python
arr1d = np.array([1, 2, 34, 13])
arr1d
````

````python
arr1d.ndim   # 1
arr1d.shape  # (4,)
````

The shape `(4,)` means the array has one axis containing 4 elements.

#### 2D array

A 2D array has two axes: rows and columns. It is usually created by passing a list of lists to `np.array()`.

Each inner list represents one row. For a proper 2D numeric array, all rows should have the same number of elements.

````python
arr2d = np.array([[1, 22, 334],
                  [3, 35, 345]])
arr2d
````

````python
arr2d.ndim  # 2
arr2d.shape #(2,3)
````

The shape `(2, 3)` means the array has 2 rows and 3 columns.

#### 3D array

A 3D array has three axes. You can think of it as a stack of 2D arrays.

It is created using nested lists with one more level of nesting.

````python
arr3d = np.array([[[1, 2, 3],
                   [12, 23, 43]],
                  [[12, 23, 2],
                   [343, 322, 3]]])
arr3d
````

````python
arr3d.ndim  #3
arr3d.shape # (2, 2, 3)
````

The shape `(2, 2, 3)` means the array has:

- 2 blocks
- 2 rows in each block
- 3 columns in each row

#### Important rule: regular shape is expected

For a normal numeric NumPy array, the nested sequence should be rectangular.

For example, this is not a proper regular 2D array:

```python
np.array([[13, 23, 434, 12],
           [12, 23, 45]])
```

The first inner list has length 4, but the second inner list has length 3. Because of this mismatch, NumPy cannot form a clean 2D numeric array.

````python
# arr = np.array([[13, 23, 434, 12],
#                 [12, 23, 45]])

# The inner list [13, 23, 434, 12] has length 4
# The inner list [12, 23, 45] has length 3
````

- 1D array shape looks like `(n,)`
- 2D array shape looks like `(rows, columns)`
- 3D array shape looks like `(depth, rows, columns)`
- NumPy works best with rectangular nested lists
