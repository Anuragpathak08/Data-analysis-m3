### Attributes of NumPy Array

Array attributes are built-in properties of a NumPy `ndarray` that describe the array itself.

They are useful for understanding:

- how many dimensions the array has
- what data type it stores
- its shape and size
- how much memory it is using

````python
import numpy as np
````

````python
arr_1 = np.array([123, 45, 23, 132, 4])

arr_2 = np.array([[12, 3, 4],
                  [12, 3, 4],
                  [3, 5, 46],
                  [2, 3, 36]])

arr_3 = np.array([[12, 3, 4, 4],
                  [12, 3, '4', 4565],
                  [3, 5, 46, 54]])

arr_4 = np.array([[[8377, 5984, 3609],
                   [5797, 683, 9525],
                   [2751, 9512, 7689],
                   [2455, 7102, 4184]],

                  [[634, 8502, 7840],
                   [2241, 2642, 1075],
                   [3114, 8.20, 548],
                   [1229, 4335, 5328]]])
````

Here we create four arrays:

- `arr_1`: a 1D array
- `arr_2`: a 2D array
- `arr_3`: a 2D array containing a string, so NumPy will use a text dtype
- `arr_4`: a 3D array containing a float, so NumPy will use a float dtype

### Attributes of Array

#### ndim

`syntax: variable.ndim`

`ndim` returns the number of dimensions, also called axes, of the array.

- 1D array has `ndim = 1`
- 2D array has `ndim = 2`
- 3D array has `ndim = 3`

````python
arr_1.ndim # 1
arr_2.ndim # 2
arr_3.ndim # 2
arr_4.ndim # 3
````

#### dtype

`syntax: variable.dtype`

`dtype` tells us the data type of the elements stored in the array.

NumPy arrays usually have a single common dtype for all elements.

````python
arr_1.dtype # dtype('int64') 
arr_2.dtype # dtype('int64')
arr_3.dtype # dtype('<U21')
arr_4.dtype # dtype('float64')
````

Here:

- `arr_1` contains only integers, so its dtype is integer
- `arr_2` contains only integers, so its dtype is integer
- `arr_3` contains the string `'4'`, so NumPy converts the array to a Unicode text dtype and becomes a string/text dtype.
- `arr_4` contains `8.20`, so NumPy promotes the array to `float64`

#### shape

`syntax: variable.shape`

`shape` returns a tuple showing the size of each axis.

For example:

- 1D shape: `(n,)`
- 2D shape: `(rows, columns)`
- 3D shape: `(blocks, rows, columns)`

````python
arr_1.shape # (5,)
arr_2.shape # (4, 3)
arr_3.shape # (3, 4)
arr_4.shape # (2, 4, 3)
````

So:

- `arr_1` has 5 elements in one dimension
- `arr_2` has 4 rows and 3 columns
- `arr_3` has 3 rows and 4 columns
- `arr_4` has 2 blocks, 4 rows, and 3 columns

#### size

`syntax: variable.size`

`size` returns the total number of elements in the array.

It is equal to the product of the shape values.

````python
arr_1.size # 5
arr_2.size # 12
arr_3.size # 12
arr_4.size # 24
````

For example, `arr_4` has shape `(2, 4, 3)`, so its size is:

```python
2 * 4 * 3 = 24
```

#### itemsize

`syntax: variable.itemsize`

`itemsize` returns the number of bytes used to store one element in the array.

The correct attribute name is `itemsize`.

````python
arr_1.itemsize # 8 
arr_2.itemsize # 8
arr_4.itemsize # 8
````

#### nbytes

`syntax: variable.nbytes`

`nbytes` returns the total number of bytes consumed by the array data.

Formula:

```python
nbytes = size * itemsize
```

````python
arr_1.nbytes # 40
arr_2.nbytes # 96
arr_4.nbytes # 192
````

For `arr_4`:

- size = 24
- itemsize = 8 bytes
- nbytes = 24 * 8 = 192 bytes

The memory usage of `arr_3` depends on its Unicode string itemsize.

- `ndim` gives the number of dimensions
- `dtype` gives the element data type
- `shape` gives the size along each axis
- `size` gives the total number of elements
- `nbytes` gives the total memory used by the array data
- `itemsize` gives the memory used by one element
