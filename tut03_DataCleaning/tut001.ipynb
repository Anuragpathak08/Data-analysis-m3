```python
titanic = pd.read_csv(r'..\dataset\titanic.csv')
```

```python
titanic.info()
```

```python
titanic.head(2)
```

```python
titanic = titanic.set_index('PassengerId')
```

### Renaming the column

```python
new_col_names = {'SibSp': 'Sibling_spouse',
                 'Parch': 'Parent_child',
                 'Embarked': 'On_board'}

titanic = titanic.rename(columns= new_col_names )
```

### Removing the duplicates

```python
# method duplicate
titanic.duplicated()
titanic.duplicated().sum()
```

```python
# method drop_duplicates
titanic = titanic.drop_duplicates()
```

```python

titanic.duplicated(subset=['Name', 'Ticket']).sum()
titanic = titanic.drop_duplicates(subset=['Name', 'Ticket'] )
```

### changing the data in a column

```python
titanic['Pclass'].unique()
```

```python
change_value = {'3rd':'3rd', 
                '1st':'1st',
                '1':'1st',
                '2':'2nd',
                '3':'3rd'}

titanic['Pclass'] = titanic['Pclass'].map(change_value)
```

```python
titanic['Pclass'].unique()
```

```python
titanic['Survived'] = titanic['Survived'].astype('bool')
```

```python
titanic.info()
```

```python
titanic['Age'].unique() 
titanic['Fare'].unique() 
titanic['Sibling_spouse'].unique()  
titanic['Parent_child'].unique() 
```

```python
titanic['Parent_child'] = titanic['Parent_child'].astype('uint8')
titanic['Sibling_spouse'] = titanic['Sibling_spouse'].astype('uint8')
```

```python
titanic.info()
```

### Handle the nan

```python
titanic['On_board'].isnull().sum()
```

```python
titanic = titanic.dropna(subset=['On_board'])
```

```python
titanic.info()
```

```python
median_age = titanic['Age'].median()
titanic['Age'] = titanic['Age'].fillna(median_age)
```

```python
titanic.info()
```

```python
titanic.Age = titanic.Age.round()
```

```python
titanic.sample()
```

### Removing the unwanted column

```python
# titanic.drop('Cabin', axis=1)
titanic = titanic.drop(columns=['Cabin'])
```

```python
titanic.info()
```
