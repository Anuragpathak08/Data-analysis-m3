```python
titanic = pd.read_csv(r'..\dataset\titanic.csv')
```

```python
titanic.info()
```

```python
titanic = titanic.set_index('PassengerId')
```

### Renaming the column

```python
new_col_names = {'SibSp': 'Sibling_spouse',
                 'Parch': 'Parent_child',
                 'Embarked': 'On_board'}

titanic = titanic.rename(columns= new_col_names )
```

### Removing the duplicates

```python
# method duplicate
titanic.duplicated()
titanic.duplicated().sum()
```

```python
# method drop_duplicates
titanic = titanic.drop_duplicates()
```

```python
titanic.duplicated(subset=['Name', 'Ticket']).sum()

titanic = titanic.drop_duplicates(subset=['Name', 'Ticket'] )
```

### changing the data in a column

```python
titanic['Pclass'].unique()
```

```python
change_value = {'3rd':'3', 
                '1st':'1',
                '1':'1',
                '2':'2',
                '3':'3'}

titanic['Pclass'] = titanic['Pclass'].map(change_value)
```
