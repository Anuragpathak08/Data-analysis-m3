### Pandas Series

A **Series** is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). It is essentially a single column of a DataFrame. The axis labels are collectively referred to as the **index**.

| Feature |  NumPy Array | Pandas Series |
| --- | --- | --- |
| **Index** | Implicit integer index only (`0, 1, 2...`) | Explicit labeled index (can be numbers, strings, or dates) |
| **Data Types** | NumPy arrays require a single homogeneous type | Homogeneous data type per Series, with specialized types (Category, Datetime) |
| **Statistical & Analysis Tools** | Basic mathematical operations (NumPy) | Rich built-in analytical methods (`.describe()`, `.value_counts()`) |
| **Memory & Performance** | NumPy is faster and uses less memory for raw math | Adds slight memory overhead to support indexes, labels, and metadata |

#### Creating a Series

You can create a Series from various Python objects:
- **Lists/Tuples**: The index will be automatically generated as integers starting from 0.
- **Dictionaries**: The keys become the index, and the values become the data.
- **Scalars**: If a scalar value is passed, it is repeated to match the length of the index (if provided) or creates a single-element Series.
- **NumPy Arrays**: Can be directly converted into a Series.

*Note: Sets are not directly supported as input for `pd.Series` because they are unordered.*

```python
pd.Series([1,234,562,35])
pd.Series((1,234,562,35))
# pd.Series({1,234,562,35}) # set is not supported by pandas series
pd.Series([{12,3,54}, {13,3,4}, {132,54,7}])
pd.Series({ 'prasad': 1,
            'mahesh': 2,
            'yathart': 3})
```

```python
np.array(10.325)
```

```python
pd.Series(True)
pd.Series(10)
pd.Series(10.132)
pd.Series(10 +10j)
pd.Series('hello')
```

```python
# When a Series contains mixed data types, Pandas assigns the 'object' dtype.
pd.Series([10, 10.23, {1,32}])
```

#### Series Attributes

Pandas Series shares many attributes with NumPy arrays, allowing you to inspect the structure of your data:
- `.ndim`: Number of dimensions (always 1 for a Series).
- `.dtype`: The data type of the elements.
- `.shape`: A tuple representing the dimensions.
- `.size`: The number of elements.
- `.nbytes`: The number of bytes consumed by the elements.
- `.name`: The name of the Series (optional).

```python
s = pd.Series([13,24,35,435,4], )
s
```

```python
# numpy attribute
s.ndim
s.dtype
s.shape
s.size
s.nbytes
# s.itemsize #error not present in pd.series
```

#### Custom Indexing

By default, Pandas assigns an integer index starting at 0. However, you can specify a custom index using the `index` parameter. 
- The length of the `index` list must match the length of the data.
- You can also assign a `name` to the Series for better identification, especially when working with DataFrames (we will learn later).

```python
pd.Series([123,34,6,57,4365,2132], 
          index = [1,0,3,2,4,1],
          name = 'demo' )
```

```python
# This will raise a ValueError because the length of the index (5) does not match the length of the data (6).
pd.Series([123,34,6,57,4365,2132], 
          index = [1,0,3,2,4],
          name = 'demo' )
```

#### Series from Dictionary with Specific Index

When creating a Series from a dictionary, you can pass an `index` argument. 
- If a label in the `index` exists in the dictionary, its value is used.
- If a label in the `index` is **not** found in the dictionary, the resulting value will be `NaN` (Not a Number), indicating missing data.

```python
pd.Series({ 'Mayuri': 1,
            'Shravani': 2,
            'Bhargavi': 3,
            'b': 3,
            'Rajnandini': 5},
          index= ['a', 'b', 'c', 'd'])
```
