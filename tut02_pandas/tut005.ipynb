```python
import numpy as np
import pandas as pd
import sqlite3
```

```python
with sqlite3.connect(r'..\dataset\books.sqlite') as conn:
    cur = conn.cursor()
    tabels = cur.execute('''SELECT name 
                            FROM sqlite_schema 
                            WHERE type = 'table' 
                            AND name NOT LIKE 'sqlite_%';''')
    print(list(tabels))
conn.close()
```

```python
with sqlite3.connect(r'..\dataset\books.sqlite') as conn:
    cur = conn.cursor()
    books = pd.read_sql('select * from books', conn)
    customers = pd.read_sql('select * from customers', conn)
```

## Methods of DataFrame

#### head<hr>
`syntax - variable.head(n)`

```python
books.head() # By default it give top 5
books.head(6)
books.head(1)
```

#### tail<hr>
`syntax - variable.tail(n)`

```python
customers.tail() # By default it give last 5
customers.tail(23)
customers.tail(3000)
customers.tail(3)
```

#### sample<hr>
`syntax : variable.sample(n)`

```python
books.sample() # default gives only one random value
books.sample(2)
books.sample(10)
```
