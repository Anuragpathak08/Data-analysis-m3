### Accessing Data in a DataFrame

*Note: While this works, Pandas discourages "chained indexing" for assignment because it can lead to unpredictable behavior (modifying a copy instead of the original). For selection, it is generally safe, but `.loc` or `.iloc` are preferred.*

```python
crime_data['Category']
# crime_data[0] # Error : only you can pass column name
crime_data['OffenderStatus']
crime_data['Offender_Age']
```

#### 2. Fancy Indexing
It is used to select multiple columns, pass a **list of column names** inside the square brackets. This returns a new DataFrame containing only those columns.

```python
america[['city', 'state', 'zipcode']]
# america[[1,3,0]] # Error : only you can pass column names 
america[['city', 'state', 'county']]
america[['city', 'state']]
```

You can slice a DataFrame directly using `[start:stop:step]`. This performs **positional slicing** on the rows.

### Advanced Indexing: `.loc` and `.iloc`

To avoid the pitfalls of chained indexing, Pandas provides two primary, robust indexers:
1. **`.loc`**: **Label-based** indexing. You use the actual row and column names/labels.
2. **`.iloc`**: **Integer-position-based** indexing. You use the numerical index (0, 1, 2...), regardless of the actual labels.

```python
df10 = pd.DataFrame([[1,3,542,43],
                    [1,233,44,35],
                    [13,2,35,46]], index = ['A', 'B', 'C'], columns=['w', 'x', 'y', 'z'] )
df10
```

#### The Danger of Chained Assignment
Attempting to assign a value using chained indexing (e.g., `df10[['w','z']][0:2:] = 100`) triggers a `ChainedAssignmentError` in modern Pandas. 

**Why?** Because `df10[['w','z']]` creates a temporary intermediate DataFrame (a copy). Assigning `= 100` modifies that temporary copy, not the original `df10`. Pandas' Copy-on-Write (CoW) mechanism explicitly blocks this to prevent silent failures.

```python
# df10[0]
df10['x']
df10['w']
df10[0:2:]
df10['A':'C':]
df10[['w','z']][0:2:] = 100
```

As seen below, the original `df10` remains completely unchanged because the assignment failed.

```python
df10
```

#### The Correct Way: Using `.loc`
`.loc[row_label, column_label]` allows you to select and **assign** values in a single, safe step. 
- `'B':` means "from row label 'B' to the end".
- `'y':` means "from column label 'y' to the end".
This successfully updates the original DataFrame.

```python
df10.loc['B']
df10.loc[['B', 'A'], ['w', 'y']]
df10.loc['B': , 'y':]
df10.loc['B': , 'y':] = [[10, 20], [30,40]]
```

Notice how rows 'B' and 'C' for columns 'y' and 'z' have been successfully updated to the new values.

```python
df10
```

#### Using `.iloc` for Positional Indexing
`.iloc[row_position, column_position]` uses strictly integer-based indexing, similar to standard Python list indexing.
- `[0, 2]` for rows means: select the 1st row (index 0) and the 3rd row (index 2).
- `[1, 0]` for columns means: select the 2nd column (index 1, which is 'x') and the 1st column (index 0, which is 'w').

```python
df10.iloc[[0, 2], [1, 0]]
```
