### Loading Data into DataFrames

The `pd.read_csv()` function is used to load comma-separated values (CSV) files into a Pandas DataFrame. It can read files from:
1. **Local file paths**: e.g., `'../dataset/crime_data.csv'`
2. **Web URLs**: e.g., a raw GitHub link to a CSV file.

```python
crime_data = pd.read_csv('../dataset/crime_data.csv')
crime_data
```

```python
america = pd.read_csv(r'https://raw.githubusercontent.com/scpike/us-state-county-zip/master/geo-data.csv')
america
```

### Accessing Data in a DataFrame

#### 1. Indexing (Column then Row)
You can access a specific value by first selecting the column (which returns a Series) and then selecting the index position. 
*Note: While this works, Pandas discourages "chained indexing" for assignment because it can lead to unpredictable behavior (modifying a copy instead of the original). For selection, it is generally safe, but `.loc` or `.iloc` are preferred.*

```python
crime_data['Category']
# crime_data[0] # Error : only you can pass column name
crime_data['OffenderStatus']
crime_data['Offender_Age']
```

```python
crime_data['Disposition'][1]
```

```python
type(crime_data['Category']) # indexing gives pandas.Series
```

```python
crime_data['Category'][2000]
crime_data['Category'][200]
crime_data['Category'][20]
crime_data['Category'][1000]
crime_data['Category'][1]
```

```python
america['city']
america['city'][34]
```

#### 2. Fancy Indexing
It is used to select multiple columns, pass a **list of column names** inside the square brackets. This returns a new DataFrame containing only those columns.

```python
america[['city', 'state', 'zipcode']]
# america[[1,3,0]] # Error : only you can pass column names 
america[['city', 'state', 'county']]
america[['city', 'state']]
```

#### 3. Slicing Rows Directly
You can slice a DataFrame directly using `[start:stop:step]`. This performs **positional slicing** on the rows. 

```python
# crime_data[-1] # error
crime_data[-1:-10:-1]
```

```python
america = america.set_index('city')
america
```

```python
df = america[0:20:1]
df['Keystone' : 'Zcta 35022']
```

### `.loc` and `.iloc`

To avoid the pitfalls of chained indexing, Pandas provides two primary, robust indexers:
1. **`.loc`**: **Label-based** indexing. 
2. **`.iloc`**: **Integer-position-based** indexing. 

```python
df10 = pd.DataFrame([[1,3,542,43],
                    [1,233,44,35],
                    [13,2,35,46]], index = ['A', 'B', 'C'], columns=['w', 'x', 'y', 'z'] )
df10
```

#### `.loc`
`.loc[row_label, column_label]` allows you to select and **assign** values.

```python
df10.loc['B']
df10.loc[['B', 'A'], ['w', 'y']]
df10.loc['B'::1 , 'y'::1]
```

```python
df10.loc['B'::1 , 'y':] = [[10, 20], [30,40]] # assign the value
df10
```

`.iloc[row_position, column_position]` uses strictly integer-based indexing, similar to standard numpy *indexing / fancy indexing / slicing.*

```python
df10.iloc[[0, 2], [1, 0]]
```
