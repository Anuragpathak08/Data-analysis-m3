### Accessing Elements in a Series

You can access elements in a Pandas Series using square brackets `[]`. The behavior depends on the type of index:
- **Integer Index (Default)**: You can use integer positions (0, 1, 2...) just like a list or NumPy array.
- **Custom Label Index**: If you have defined custom labels (strings, floats, etc.), you must use those specific labels to access the data.

```python
s1 = pd.Series([13,22,54,2,43,5])
s1
```

syntax : variable[index]

```python
s1[0]
s1[2]
s1[3]
# s1[10] # error coz no 10th value
# s1[-2]
```

```python
s2 = pd.Series([13,2,345,6,34,24], index=[1,6,5,4,3,2])
s2
```

```python
# When using a custom integer index, s2[1] looks for the label '1', not the first position.
s2[1]
s2[2]
```

```python
s3 = pd.Series([22, 24, 12, 8], index = ['om', 'bob', 'tom', 'sheru'])
s3
```

```python
# s3[0] #error - Integer positional access is often restricted when a non-integer index is present.
s3['om']
```

### Slicing

Slicing allows you to extract a subset of the Series. 
- **Positional Slicing**: Works similarly to Python lists (`start:stop:step`). The `stop` index is exclusive.
- **Label Slicing**: When using custom labels, both the `start` and `stop` labels are **inclusive**.

```python
s1
```

```python
s1[ 0: 10 : 1]
s1[ -1 : : -1]
```

```python
s2
```

```python
# Positional slicing: starts at position 1, stops before position 2.
s2[1:2:]
```

```python
s3
```

```python
s3[0:3]
# Label slicing: 'om' to 'sheru' includes both endpoints.
s3['om':'sheru']
```

```python
s4 = pd.Series([123,3421,232,465], index = [21.3, 23.2, 4.5, 24])
s4
```

```python
s4[0::]
s4[21.3::]
# s4[2.2::]
```

```python
s5 = pd.Series([13,2232,34,34], index = [1,2.23, 11.3, 12.4])
s5
```

```python
# Slicing with float labels: Pandas finds the position of 10.3 and slices from there.
s5[10.3::]
```

### Boolean Indexing (Filtering)

You can filter a Series by passing a boolean condition inside the square brackets. This returns only the elements where the condition is `True`.

```python
s1
```

```python
s1[s1 % 2 == 0]
s1[(s1%2==0) | (s1%3==0)]
```

```python
s3
```

```python
# Filtering based on index labels (names starting with a vowel)
l = []
for i in s3.index:
    l.append(i[0] in 'AEIOUaeiou') 

s3[l]
```

```python
# Filtering using a manual boolean list
s3[[True, True, False, False]]
```

### Series Properties and Inspection

Pandas provides several attributes and methods to inspect the structure and content of a Series:
- `.dtype`: Data type of the values.
- `.shape` / `.size`: Dimensions and total element count.
- `.info()`: A concise summary including memory usage and non-null counts.
- `.nbytes`: Total bytes consumed by the data.
- `.is_unique`: Checks if all values are distinct.
- `.index`: The index object of the Series.
- `.values`: The underlying NumPy array of the data.

```python
s1.dtype
s2.dtype
s3.dtype
s4.dtype
```

### Exploratory Methods

These methods help you quickly understand the distribution and content of your data:

#### `.head(n)` / `.tail(n)`
View the first or last `n` rows.

`syntax : series.head( no_of_rows )`

```python
s3 = pd.Series(np.random.randint(10,20, 30))
s3.head(6)
s3.head(2)
```

`syntax : series.tail( no_of_rows )`

```python
s3.tail(2)
s3.tail()
s3.tail(8)
```

#### `.sample(n)`: 

Return random rows.
`syntax : series.sample( no_of_rows )`

```python
s3.sample()
s3.sample(10)
```

#### `.value_counts()`
Count unique values.  `syntax : series.value_count()`

```python
s3.value_counts()
```

#### `.info`
Give the information about the series  `syntax : series.info()`

```python
s3.info()
```

#### `.describe()`
Generate descriptive statistics (mean, std, min, max, etc.). `syntax : series.describe()`

```python
s3.describe()
```

#### `.sort_values()` / `.sort_index()`: 
Sort by data values or index labels.

`syntax : series.sort_values(ascending = True)`

```python
s3.sort_values()
```

`syntax : series.sort_index(ascending = True)`

```python
s3.sort_index()
```
